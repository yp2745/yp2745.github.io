### [Reading an Excel file](https://pandas.pydata.org/docs/user_guide/io.html#excel-files)

Worst case, you can convert the data to CSV and read that.

- [Excel](https://support.microsoft.com/en-us/office/import-or-export-text-txt-or-csv-files-5250ac4c-663c-47ce-937b-339e391393ba)
- [Google Sheets](https://support.ecwid.com/hc/en-us/articles/8578742087580-Opening-and-saving-CSV-files-in-Google-Sheets)


```python
import pandas as pd

happiness = pd.read_excel(
    "World_Happines_Report.xlsx", sheet_name="Data for Figure 2.1 (2011–2025)"
)
happiness.head()

electricity = pd.read_csv("Access_to_Electricity.csv", skiprows=4)
electricity.head()
```

```python
# Select China as target country for electricity
indonesia_electricity = electricity[electricity["Country Name"] == "Indonesia"]
indonesia_electricity
```

```python
# Select China as target country for happiness
indonesia_happiness = happiness[happiness["Country name"] == "Indonesia"]
indonesia_happiness
```

```python
# Wide to long form
years = [str(i) for i in range(2011, 2026)]

indonesia_electricity = indonesia_electricity.melt(
    id_vars="Country Name", value_vars=years, var_name="Year", value_name="Electricity"
)

indonesia_electricity["Year"] = indonesia_electricity["Year"].astype(int)

indonesia_electricity.head()
```

```python
# Merge two datasets
indonesia = pd.merge(indonesia_electricity, indonesia_happiness, on="Year", how="inner")

indonesia.head()
```

```python
# Draw graph
import plotly.express as px

fig = px.scatter(
    indonesia,
    x="Electricity",
    y="Life evaluation (3-year average)",
    color="Year",
    title="Electricity Access vs. Happiness in Indonesia",
)

fig.show()
```

```python
# Two Y-axises
import plotly.graph_objects as go
from plotly.subplots import make_subplots

fig = make_subplots(specs=[[{"secondary_y": True}]])

# Electricity - left Y-axis
fig.add_trace(
    go.Scatter(
        x=indonesia["Year"],
        y=indonesia["Electricity"],
        name="Electricity",
        mode="lines+markers",
    ),
    secondary_y=False,
)

# Life Evaluation - right Y-axis
fig.add_trace(
    go.Scatter(
        x=indonesia["Year"],
        y=indonesia["Life evaluation (3-year average)"],
        name="Life Evaluation",
        mode="lines+markers",
    ),
    secondary_y=True,
)

fig.update_xaxes(title_text="Year")
fig.update_yaxes(title_text="Electricity Access (%)", secondary_y=False)
fig.update_yaxes(title_text="Life Evaluation", secondary_y=True)

fig.update_layout(
    title="Electricity Access and Happiness in Indonesia", template="plotly_white"
)

fig.show()
```
