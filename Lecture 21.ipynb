```python
import pandas as pd

df = pd.read_csv("WDICSV.csv")
df.head()
```

```python
electricity = df[df["Indicator Code"] == "EG.ELC.ACCS.ZS"].copy()
electricity = electricity.iloc[49:].copy()
electricity
```

```python
import plotly.express as px

data = electricity.set_index("Country Name").loc[:, "1990":"2024"].T

fig = px.line(data, title="Access to Electricity by Country Over Time")

fig.update_layout(
    xaxis_title="Year", yaxis_title="Access to Electricity (%)", showlegend=False
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px

years = [
    col for col in electricity.columns if col.isdigit() and 1990 <= int(col) <= 2024
]

data = electricity.melt(
    id_vars=["Country Name", "Country Code"],
    value_vars=years,
    var_name="Year",
    value_name="Access",
)

data["Access"] = pd.to_numeric(data["Access"], errors="coerce")
data = data.dropna(subset=["Access"])

fig = px.choropleth(
    data,
    locations="Country Code",
    locationmode="ISO-3",
    color="Access",
    hover_name="Country Name",
    animation_frame="Year",
    color_continuous_scale="Viridis",
    range_color=(0, 100),
    title="Access to Electricity by Country Over Time",
)

fig.update_layout(coloraxis_colorbar_title="Access (%)")

fig.show()
```
