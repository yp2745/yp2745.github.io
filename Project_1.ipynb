## Load Data without Pandas

Let's first load data again without Pandas. The score column in this dataset contains many missing values, so I will skip the missing record to make the results consistent with the previous method.

```python
import csv

scores = []

with open(
    "DOHMH_New_York_City_Restaurant_Inspection_Results_20261112.csv", "r"
) as file:
    restaurants = csv.DictReader(file)

    for row in restaurants:
        if row["SCORE"] != "":
            scores.append(int(row["SCORE"]))
```

## Analysis (Hard way)

Awesome! We have loaded the data. Let's calculate the mean, median, and mode again without using Pandas and see if we get the same results.

### Mean

To calculate mean in a hard way, I will sum up all the scores and divided by the number of scores.

```python
sum(scores) / len(scores)
```

We got the same mean!

### Median

To find the median, I will sort all the values in ascending order and find the middle value. If the dataset has an even number of scores, I will add the two middle values together and divide by 2.

```python
sorted_scores = sorted(scores)

if len(sorted_scores) % 2 == 1:
    median = sorted_scores[len(sorted_scores) // 2]
else:
    middle1 = sorted_scores[len(sorted_scores) // 2 - 1]
    middle2 = sorted_scores[len(sorted_scores) // 2]
    median = (middle1 + middle2) / 2

print(median)
```

We got the same median as well!

### Mode

For mode, I will count the number of times each unique score appears and identify the score with the highest frequency.

```python
counts = {}

for i in scores:
    if i in counts:
        counts[i] += 1
    else:
        counts[i] = 1

mode = []

for i in counts:
    if counts[i] == max(counts.values()):
        mode.append(i)

print(mode)
```

Awesome! Even we used two different methods, we got the same mean, median, and mode!

## Data Visualization

With thousands of rows of scores, it is difficult to understand the overall food safety conditions in NYC just by looking at the numbers, so we now try some data visualizations to make the data look nicer. 

According to NYC Health, scores of 0-13 points will be identified as satisfactory conditions and will be given a grade A; 14-27 points mean there are critical issues found in the restaurant that need fixing and will be given a grade B; 28+ points mean there are significant public health concerns and will be granted a grade C. 

For our purposes, I will group all the inspection scores into 3 categories (grade A, grade B, and grade C), and see how many records fall into each category.

```python
grade_A = 0
grade_B = 0
grade_C = 0

for i in scores:
    if i <= 13:
        grade_A += 1
    elif i <= 27:
        grade_B += 1
    else:
        grade_C += 1

print("NYC Restaurant Inspection Score Distribution")
print("Score Range          Number of Records")

largest = max(grade_A, grade_B, grade_C)

print(f"Grade A (0-13): {'*' * round(grade_A / largest * 30)}")
print(f"Grade B (0-13): {'*' * round(grade_B / largest * 30)}")
print(f"Grade C (28+): {'*' * round(grade_C / largest * 30)}")
```

Now, it is very easy to see that among all records, Grade A is the most common category (nice!), followed by Grade C and Grade B. However, the difference in frequency between Grade A and Grade C is not very large.
